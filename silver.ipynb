# Convert to titlecase - initcap

```python
df = df.withColumn("Country",initcap(df["Country"]))
df = df.withColumn("ProductCategory",initcap(df["ProductCategory"]))
df = df.withColumn("ProductName",initcap(df["ProductName"]))
df = df.withColumn("SalesRegion",initcap(df["SalesRegion"]))
df = df.withColumn("CustomerName",initcap(df["CustomerName"]))

df.display()
```

# Remove rows where all of the values are mising

```python
df = df.dropna(how='all')
df.display()
```

# Remove duplicate rows 

```python
df = df.dropDuplicates()
df.display()
```

# Move the data into Silver Layer 

```python
df.write.format('parquet').mode('overwrite').option("path","abfss://silver@cfstorageaccountrkd.dfs.core.windows.net/silverdata").save()
```
